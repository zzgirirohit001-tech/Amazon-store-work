```python
#AQI Data set 
df = pd.read_csv("globalAirQuality.csv")
df.head()
df.tail()
df.max()
df.min()
```

```python
#selecting the data 
df["city"]
df[["city","aqi"]]

df.loc[1]
df.loc[0:2]  #satrt index : end index (inclusive)

df.iloc[2]
df.iloc[0:2] # end index exclusive

df.iloc[0:2 , 2]
df.loc[0:2 , ["city","aqi","latitude","longitude"]]
```

```python
#Filtering of data
df[df["aqi"] > 100]
df[(df["aqi"] > 100)&(df["temperature"] > 30) ]
df[df["aqi"] > 100][["city","aqi"]]
```

```python
#Query
aqi_val =100
df.query("aqi >100 & temperature > 30")
df.query("aqi >@aqi_val & temperature > 30")
aqi_data = df.query("aqi >@aqi_val & temperature > 30").copy()
aqi_data
```

```python
#Cleaning raw data - missing values
df = pd.read_csv("raw_data.csv")
df.isnull()

df.isnull().sum()

df.dropna() # drop rows
df.dropna(axis = 1) # drop column

df.fillna(0)

df["age"].fillna(df["age"].mean())

df.ffill()
df.bfill()
```

```python
#Handle duplicates

df.duplicated()
df[["country","gender"]].duplicated()
df.drop_duplicates()
```
