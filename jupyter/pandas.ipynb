```python
#Grouping and Aggregation of Data
df.groupby("country")["income"].mean()
df.groupby("country")["income"].max()
df.groupby("country")["income"].min()

df.groupby("gender")["income"].mean()
df.groupby("gender")["income"].max()
df.groupby("gender")["income"].min()

#calculate all the values in one line 
df.groupby("country")["income"].agg(["mean","min","max"])

#passing new value or new name to the three parameters
df.groupby("country")["income"].agg(avg_salary = "mean",min_salary = "min",max_salary = "max")

```

```python
#want to change the format of table line wide to short by changing the row and column
df = pd.DataFrame({
    "country" : ["USA","USA", "India","India"],
    "year" : [2020,2021,2020,2021],
    "sales" : [100,120,300,400],
    "profit" : [20,25,100,200],
    "tax" : [10,20,30,40]
})

melted_df = df.melt(
    id_vars = ["country","year"],
    value_vars = ["sales","profit"],
    var_name = "matrics",
    value_name = "value"
)

original = melted_df.pivot(
    index=["country", "year"],
    columns="matrics",
    values="value"
)
original
# print(melted_df.columns)
# melted_df
```
