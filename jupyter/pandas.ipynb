```python
#AQI Data set 
df = pd.read_csv("globalAirQuality.csv")
df.head()
df.tail()
df.max()
df.min()
```

```python
#selecting the data 
df["city"]
df[["city","aqi"]]

df.loc[1]
df.loc[0:2]  #satrt index : end index (inclusive)

df.iloc[2]
df.iloc[0:2] # end index exclusive

df.iloc[0:2 , 2]
df.loc[0:2 , ["city","aqi","latitude","longitude"]]
```

```python
#Filtering of data
df[df["aqi"] > 100]
df[(df["aqi"] > 100)&(df["temperature"] > 30) ]
df[df["aqi"] > 100][["city","aqi"]]
```

```python
#Query
aqi_val =100
df.query("aqi >100 & temperature > 30")
df.query("aqi >@aqi_val & temperature > 30")
aqi_data = df.query("aqi >@aqi_val & temperature > 30").copy()

```
