```python
# Load the data
def load_data(filename):
    with open(filename, 'r') as f:
        data =json.load(f)
    return data
```

```python
# def clean_data(data):
#     text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
#     clean_data=[]
#     unique_user= set()
#     for user in data:
#       raw_rating=str(user["rating"]).strip().lower()
#       if(raw_rating in text_to_num):
#           raw_rating = text_to_num[raw_rating]
#       print(raw_rating)
#       user["rating"]= raw_rating
#         #handel missing value
#       raw_age = user.get("age")
#       if(raw_age==None):
#             user["age"] = None
#       if user["name"].strip() in unique_user:
#           continue
#           clean_data.append(user)
#     return clean_data
def clean_data(data):
    text_to_num = {
        "one": 1,
        "two": 2,
        "three": 3,
        "four": 4,
        "five": 5
    }

    clean_data = []
    unique_user = set()

    for user in data:

        raw_rating = str(user["rating"]).strip().lower()

        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]

        user["rating"] = raw_rating

        # handle missing value
        raw_age = user.get("age")

        if raw_age == None:
            user["age"] = None

        # duplicate user check
        name = user["name"].strip()

        if name in unique_user:
            continue

        unique_user.add(name)

        # add cleaned user
        clean_data.append(user)

        # print all details
        print(user)

    return clean_data
```

```python
data = clean_data(data)
```

```python
#get meaningful insight from data
def get_insight(data):
    #avg_raitng
    total_rating = 0
    for user in data:
        total_rating+=float(user["rating"])
    print(f"avg rating = {total_rating/len(data)}")
    #percentage of user with poor rating
    poor_ratings = 0
    for user in data:
        if(float(user["rating"])<3):
            poor_ratings += 1
    print(f"Percentage of poor rating of user  = {poor_ratings/len(data)*100}")
```

```python
get_insight(data)
```

```python
# Building recommendation feature
def get_recommendation(data):
    recommendation = []

    for user in data:
        curr_recom = {}

        if float(user["rating"]) >= 4:
            curr_recom["brand"] = "Apple"
        else:
            curr_recom["brand"] = "Samsung"

        recommendation.append(curr_recom)

    return recommendation
```

```python
get_recommendation(data)
```
