```python
transactions = [
    {'customer_name': 'Mira', 'amount': 56, 'date': '2024-08-01'},
    {'customer_name': 'Mina', 'amount': 12.3, 'date': '2024-09-05'},
    {'customer_name': 'Mohan', 'amount': 21, 'date': '2024-09-03'},
    {'customer_name': 'Mohan', 'amount': 36.8, 'date': '2024-09-10'},
    {'customer_name': 'Lucy', 'amount': 66.3, 'date': '2024-09-10'}
]
```

```python
for trans in transactions:
    print(trans['customer_name'])
```

```python
#saving names in set
customers = set()
for trans in transactions:
    customers.add(trans['customer_name'])

customers
```

```python
#by tuple conprehension
customers = {trans['customer_name']for trans in transactions}
customers
```

```python
#changing in list 
customers_list = list(customers)
customers_list
```

```python
number =[1,2,3,4,5]
sqr_number = []
for n in number:
    sqr_number.append(n*n)
print(sqr_number)
```

```python
#list comprehension
sqr_numbers = [n*n for n in number]
sqr_numbers
```

```python
#checking the time taken by list and list comp
import time
long_list = range(10000000)
sqr_list = []
start = time.time()
for n in long_list:
    sqr_list.append(n*n)
end = time.time()
print("Time taken by for loop", (end-start))
```

```python
start = time.time()
sqr_list = [n*n for n in long_list]
end = time.time()
print("TIme taken by list comprehension", (end - start))
```

```python
price = [100,128,80,50]
earning_per_share = [9,15,40,2]
```

```python
pe = [p/e for p,e in zip(price , earning_per_share)]
pe
```

```python
#if earning is 0
price = [100,128,80,50]
earning_per_share = [0,15,40,2]

pe = [p/e if e!=0 else - 1 for p,e in zip(price , earning_per_share)]
pe
```
